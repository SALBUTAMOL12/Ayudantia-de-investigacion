```python
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
import torch.nn.functional as F
import tifffile as tiff
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
import ast


# 1. Definir las columnas de etiquetas
label_columns = [
    'educacion_jefe_hogar',
    'hogar_urbano',
    'mujer_jefe_hogar',
    'densidad_hogar',
    'educacion_promedio_hogar',
    'jefe_trabajo_semana_pasada',
    'material_aceptable',
    'material_recuperable',
    'dependencia_hogar',
    'numero_personas',
    'edad_promedio_hogar'
]

# 2. Split inicial: train (70%) y temp (30%), luego temp→val (15%) + test (15%)
train_df, temp_df = train_test_split(df, test_size=0.3, random_state=42)
val_df, test_df  = train_test_split(temp_df, test_size=0.5, random_state=42)

# 3. Ajustar MinMaxScaler SOLO con train_df[label_columns]
scaler = MinMaxScaler(feature_range=(0, 1))
scaler.fit(train_df[label_columns])

# 4. Transformar train, val y test SIN volver a fit
train_df[label_columns] = scaler.transform(train_df[label_columns])
val_df[label_columns]   = scaler.transform(val_df[label_columns])
test_df[label_columns]  = scaler.transform(test_df[label_columns])

# 5. Resetear índices para facilidad
train_df = train_df.reset_index(drop=True)
val_df   = val_df.reset_index(drop=True)
test_df  = test_df.reset_index(drop=True)

# 6. Definir transformación para imágenes (256×256)
class ResizeTensor(object):
    def __init__(self, size, mode='bilinear'):
        self.size = size
        self.mode = mode

    def __call__(self, tensor):
        tensor = tensor.unsqueeze(0)  # [1, C, H, W]
        tensor = F.interpolate(
            tensor, size=self.size, mode=self.mode, align_corners=False
        )
        return tensor.squeeze(0)     # [C, H, W]

transform = transforms.Compose([
    ResizeTensor((256, 256))
])

# 7. Dataset personalizado
class RegionDataset(Dataset):
    def __init__(self, df, transform=None, label_cols=None):
        self.df = df
        self.transform = transform
        self.label_cols = label_cols

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        imgs = []
        for path in row['imagenes']:
            img_np = tiff.imread(path)
            if img_np.ndim == 3 and img_np.shape[-1] == 10:
                img = torch.from_numpy(img_np).permute(2,0,1)
            elif img_np.ndim == 2:
                img = torch.from_numpy(img_np).unsqueeze(0)
            else:
                img = torch.from_numpy(img_np)
            img = img.float() / 255.0
            if self.transform:
                img = self.transform(img)
            imgs.append(img)
        images = torch.stack(imgs, dim=0)  # [n_imgs, C, H, W]

        labels = torch.tensor(
            [row[col] for col in self.label_cols],
            dtype=torch.float
        )
        return images, labels

# 8. Crear datasets
train_dataset = RegionDataset(train_df, transform=transform, label_cols=label_columns)
val_dataset   = RegionDataset(val_df,   transform=transform, label_cols=label_columns)
test_dataset  = RegionDataset(test_df,  transform=transform, label_cols=label_columns)

# 9. Collate fn para pad/truncar a 41 imágenes
def custom_collate_fn(batch, max_images=41):
    imgs_list, labels_list = zip(*batch)
    padded = []
    for imgs in imgs_list:
        n, C, H, W = imgs.shape
        if n < max_images:
            pad = torch.zeros(max_images - n, C, H, W, dtype=imgs.dtype)
            imgs = torch.cat([imgs, pad], dim=0)
        else:
            imgs = imgs[:max_images]
        padded.append(imgs)
    batch_imgs = torch.stack(padded, dim=0)   # [B, 41, C, H, W]
    batch_lbls = torch.stack(labels_list, dim=0)
    return batch_imgs, batch_lbls

# 10. DataLoaders
batch_size = 8
train_loader = DataLoader(
    train_dataset, batch_size=batch_size, shuffle=True,
    collate_fn=custom_collate_fn, num_workers=8
)
val_loader = DataLoader(
    val_dataset, batch_size=batch_size, shuffle=False,
    collate_fn=custom_collate_fn, num_workers=8
)
test_loader = DataLoader(
    test_dataset, batch_size=batch_size, shuffle=False,
    collate_fn=custom_collate_fn, num_workers=8
)

# 11. Prueba de shapes
for imgs, lbls in train_loader:
    print("Batch images shape:", imgs.shape)  # [8,41,10,256,256]
    print("Batch labels shape:", lbls.shape)  # [8,11]
    break
```

```python
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
import torch.nn.functional as F
import tifffile as tiff
import pandas as pd
from sklearn.model_selection import train_test_split
import ast



# 2. Definir una transformación para redimensionar tensores usando interpolación
class ResizeTensor(object):
    def __init__(self, size, mode='bilinear'):
        """
        size: tupla (alto, ancho) deseados.
        mode: método de interpolación.
        """
        self.size = size
        self.mode = mode

    def __call__(self, tensor):
        # tensor con forma [C, H, W]
        tensor = tensor.unsqueeze(0)  # Agrega dimensión de batch: [1, C, H, W]
        tensor = F.interpolate(tensor, size=self.size, mode=self.mode, align_corners=False)
        tensor = tensor.squeeze(0)      # Vuelve a quitar la dimensión de batch
        return tensor

# Definir la transformación compuesta (redimensiona a 256x256)
transform = transforms.Compose([
    ResizeTensor((256, 256))

])

# 3. Definir el Dataset personalizado usando tifffile para cargar imágenes de 10 canales
class RegionDataset(Dataset):
    def __init__(self, df, transform=None, label_cols=None):
        """
        df: DataFrame con la columna 'imagenes' (lista de rutas de imágenes TIFF) y columnas de etiquetas.
        transform: Transformaciones a aplicar a cada imagen (en formato tensor).
        label_cols: Lista de nombres de columnas que corresponden a los labels.
        """
        self.df = df
        self.transform = transform
        self.label_cols = label_cols

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        image_paths = row['imagenes']
        images = []

        for path in image_paths:
            # Cargar la imagen con tifffile
            img_np = tiff.imread(path)
            # Si la imagen viene en forma (H, W, C) y C==10, la permutamos a (C, H, W)
            if img_np.ndim == 3 and img_np.shape[-1] == 10:
                img = torch.from_numpy(img_np).permute(2, 0, 1)
            elif img_np.ndim == 2:
                # Caso de imagen en escala de grises, se agrega dimensión de canal
                img = torch.from_numpy(img_np).unsqueeze(0)
            else:
                # Si la forma ya es (C, H, W), se asume que es correcta
                img = torch.from_numpy(img_np)

            # Convertir a float y normalizar (suponiendo valores en [0, 255])
            img = img.float() / 255.0

            # Aplicar la transformación (redimensionamiento, etc.)
            if self.transform:
                img = self.transform(img)

            images.append(img)

        # Convertir la lista de imágenes a un tensor de forma [n_imágenes, C, H, W]
        images = torch.stack(images, dim=0)

        # Extraer las etiquetas continuas de las columnas especificadas
        if self.label_cols:
            labels = torch.tensor([row[col] for col in self.label_cols], dtype=torch.float)
        else:
            labels = None

        return images, labels

# 4. Definir las columnas de etiquetas
label_columns = [
    'educacion_jefe_hogar',
    'hogar_urbano',
    'mujer_jefe_hogar',
    'densidad_hogar',
    'educacion_promedio_hogar',
    'jefe_trabajo_semana_pasada',
    'material_aceptable',
    'material_recuperable',
    'dependencia_hogar',
    'numero_personas',
    'edad_promedio_hogar'
]

# 5. Separar el DataFrame en conjuntos de entrenamiento, validación y testeo
train_df, temp_df = train_test_split(df, test_size=0.3, random_state=42)
val_df, test_df = train_test_split(temp_df, test_size=0.5, random_state=42)

train_dataset = RegionDataset(train_df, transform=transform, label_cols=label_columns)
val_dataset   = RegionDataset(val_df, transform=transform, label_cols=label_columns)
test_dataset  = RegionDataset(test_df, transform=transform, label_cols=label_columns)

# 6. Definir un collate_fn personalizado para asegurar que cada región tenga 41 imágenes
def custom_collate_fn(batch, max_images=41):
    """
    Para cada muestra (región), se paddean o truncan las imágenes a 41.
    batch: lista de tuplas (images, labels) donde images tiene forma [n_imágenes, C, H, W].
    """
    images_list, labels_list = zip(*batch)
    padded_images = []

    for images in images_list:
        n_img, C, H, W = images.shape
        if n_img < max_images:
            pad_tensor = torch.zeros(max_images - n_img, C, H, W, dtype=images.dtype)
            padded = torch.cat([images, pad_tensor], dim=0)
        elif n_img > max_images:
            padded = images[:max_images]
        else:
            padded = images
        padded_images.append(padded)

    # Forma final: [batch_size, max_images, C, H, W]
    padded_images = torch.stack(padded_images, dim=0)
    labels = torch.stack(labels_list, dim=0)

    return padded_images, labels

# 7. Crear los DataLoaders
batch_size = 8
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, collate_fn=custom_collate_fn, num_workers=8)
val_loader   = DataLoader(val_dataset, batch_size=batch_size, shuffle=False, collate_fn=custom_collate_fn, num_workers=8)
test_loader  = DataLoader(test_dataset, batch_size=batch_size, shuffle=False, collate_fn=custom_collate_fn, num_workers=8)

# Ejemplo de iteración sobre un batch
for batch_images, batch_labels in train_loader:
    print("Batch images shape:", batch_images.shape)  # Debería ser [batch_size, 41, 10, 256, 256]
    print("Batch labels shape:", batch_labels.shape)  # Debería ser [batch_size, 11]
    break
```

```python
# ── 6. TRAIN LOOP ─ (checkpoints + R²) ────────────────────────────────────
epochs        = 150
best_val_loss = float('inf')
ckpt_path     = '/content/drive/MyDrive/bigearthnet_finetune_best.pth'  # ruta modelo 1

# /content/drive/MyDrive/

for ep in range(epochs):
    #----------- TRAIN -----------
    model.train(); run = 0; p_tr = []; y_tr = []
    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        opt.zero_grad()
        out  = model(x)
        loss = criterion(out, y)
        loss.backward()
        opt.step()
        run += loss.item() * x.size(0)
        p_tr.append(out.detach().cpu()); y_tr.append(y.cpu())
    warmup(ep); cosine.step(ep + 1)

    # ---- métricas TRAIN (escala original) ----
    p_tr = torch.cat(p_tr); y_tr = torch.cat(y_tr)
    p_tr_o = torch.tensor(scaler.inverse_transform(p_tr), dtype=torch.float32)
    y_tr_o = torch.tensor(scaler.inverse_transform(y_tr), dtype=torch.float32)
    r2_tr = [r2_score(y_tr_o[:, i], p_tr_o[:, i]) for i in range(len(label_columns))]
    train_loss = run / len(train_loader.dataset)

    #----------- VALID -----------
    model.eval(); v_run = 0; p_val = []; y_val = []
    with torch.no_grad():
        for x, y in val_loader:
            x, y = x.to(device), y.to(device)
            out  = model(x)
            v_run += criterion(out, y).item() * x.size(0)
            p_val.append(out.cpu()); y_val.append(y.cpu())

    # ---- métricas VAL (escala original) ----
    p_val = torch.cat(p_val); y_val = torch.cat(y_val)
    p_val_o = torch.tensor(scaler.inverse_transform(p_val), dtype=torch.float32)
    y_val_o = torch.tensor(scaler.inverse_transform(y_val), dtype=torch.float32)
    r2_val = [r2_score(y_val_o[:, i], p_val_o[:, i]) for i in range(len(label_columns))]
    val_loss = v_run / len(val_loader.dataset)

    # ---- impresión resumen ----
    print(f"Epoch {ep+1:3d}/{epochs} | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f}")
    for col, r2t, r2v in zip(label_columns, r2_tr, r2_val):
        print(f"   R² {col:<25}: train={r2t:+.4f}   val={r2v:+.4f}")

    #----------- CHECKPOINT: guarda el mejor por Val Loss -----------
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        torch.save({
            'epoch'      : ep + 1,
            'model_state': model.state_dict(),
            'opt_state'  : opt.state_dict(),
            'sched_state': cosine.state_dict(),
            'best_val'   : best_val_loss,
            'scaler'     : scaler
        }, ckpt_path)
        print(f"   ➜ Checkpoint actualizado  (val_loss = {val_loss:.4f})")

# ---- SNAPSHOT FINAL ------------------------------------------------------
torch.save(model.state_dict(), '/content/drive/MyDrive/bigearthnet_finetune_final_weights_modelo_1.pth')
```

```python

# ╔══════════════════════════════════════════════════════════════════════════╗
# ║  BigEarthNet v2 – fine‑tuning v2                                        ║
# ║   (2)  UWLoss  homoscedástica                                           ║
# ║   (3)  Un‑freeze progresivo (layer4 → layer3)                            ║
# ║   (4)  One‑Cycle LR policy (sin warm‑up + cosine)                        ║
# ║   (6)  Data‑aug avanzado + MixUp en train                               ║
# ║   • R² en escala original                                               ║
# ╚══════════════════════════════════════════════════════════════════════════╝
import random, math, torch, tifffile as tiff
import torch.nn as nn, torch.nn.functional as F, torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import r2_score
from reben_publication.BigEarthNetv2_0_ImageClassifier import BigEarthNetv2_0_ImageClassifier

# ───────────────────────── 1. DATA‑AUG ──────────────────────────────────────
class GeoAug:                               # flips + 90° rot
    def __call__(self,x):
        if random.random()<.5: x=torch.flip(x,[2])
        if random.random()<.5: x=torch.flip(x,[1])
        k=random.randint(0,3)
        return torch.rot90(x,k,[1,2]) if k else x

class RadJitter:                            # brillo/contraste aleatorio
    def __init__(self,alpha=.15,beta=.15): self.a,self.b=alpha,beta
    def __call__(self,x):
        g = 1.+(random.random()*2-1.)*self.a
        b = (random.random()*2-1.)*self.b
        return torch.clamp(x*g + b, 0., 1.)

class RandResizedCrop:
    def __init__(self,size,scale=(.8,1.)):
        self.size,self.scale=size,scale
    def __call__(self,x):                   # x [C,H,W] – crop centrado rand
        C,H,W = x.shape
        s = random.uniform(*self.scale)
        h = int(H*s); w = int(W*s)
        top = random.randint(0,H-h); left=random.randint(0,W-w)
        x = x[:,top:top+h,left:left+w]
        return F.interpolate(x.unsqueeze(0),self.size,
                             mode='bilinear',align_corners=False).squeeze(0)

def aug_train(x):
    x = RandResizedCrop((256,256))(x)
    x = GeoAug()(x)
    x = RadJitter()(x)
    return x
aug_val  = lambda x: F.interpolate(x.unsqueeze(0),(256,256),
                                   mode='bilinear',
                                   align_corners=False).squeeze(0)

# reasignamos
train_dataset.transform = aug_train
val_dataset.transform   = aug_val
test_dataset.transform  = aug_val

# ───────────────────────── 2. BACKBONE ─────────────────────────────────────
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
backbone = BigEarthNetv2_0_ImageClassifier.from_pretrained(
          'BIFOLD-BigEarthNetv2-0/resnet50-s2-v0.2.0')
backbone.model.vision_encoder.fc = nn.Identity()
backbone.to(device).eval()

for p in backbone.parameters(): p.requires_grad=False
for n,p in backbone.model.vision_encoder.named_parameters():
    if 'layer4' in n: p.requires_grad=True                # solo layer4 al inicio

# ───────────────────────── 3. HEAD ─────────────────────────────────────────
class FineTuneHead(nn.Module):
    def __init__(self,out_dim=11,hid=1024,p=0.3):
        super().__init__()
        self.net=nn.Sequential(nn.Linear(2048,hid), nn.LayerNorm(hid), nn.ReLU(),
                               nn.Dropout(p),
                               nn.Linear(hid,hid//2), nn.LayerNorm(hid//2),
                               nn.ReLU(), nn.Dropout(p),
                               nn.Linear(hid//2,out_dim))
    def forward(self,x): return self.net(x)

class Model(nn.Module):
    def __init__(self,bb): super().__init__(); self.bb=bb; self.hd=FineTuneHead()
    def forward(self,imgs):                      # [B,S,C,H,W]
        B,S,C,H,W = imgs.shape
        f=self.bb(imgs.view(B*S,C,H,W)).view(B,S,-1).mean(1)
        return self.hd(f)

model = Model(backbone).to(device)

# ───────────────────────── 4. UWLoss (2) ───────────────────────────────────
class UWLoss(nn.Module):
    def __init__(self,k): super().__init__(); self.log_var=nn.Parameter(torch.zeros(k))
    def forward(self,ŷ,y):
        prec=torch.exp(-self.log_var)
        mse=(ŷ-y).pow(2).mean(0)
        return .5*((prec*mse)+self.log_var).sum()

criterion = UWLoss(11).to(device)

# ───────────────────────── 5. OPTIM & ONE‑CYCLE (4) ────────────────────────
lr_back  = 1e-4
lr_head  = 5e-4
wd = 1e-4

opt = optim.AdamW(
        [{'params':[p for p in backbone.parameters() if p.requires_grad], 'lr':lr_back},
         {'params':model.hd.parameters(),                              'lr':lr_head},
         {'params':criterion.parameters(),                             'lr':lr_head}],
         weight_decay=wd)

steps_per_epoch = len(train_loader)
scheduler = optim.lr_scheduler.OneCycleLR(opt, max_lr=[lr_back,lr_head,lr_head],
                                         steps_per_epoch=steps_per_epoch,
                                         epochs=60, pct_start=.05,
                                         div_factor=25, final_div_factor=1e4)

# ───────────────────────── 6. MIXUP (6) ────────────────────────────────────
def mixup(x,y,alpha=.4):
    λ = torch.distributions.Beta(alpha,alpha).sample().item()
    idx = torch.randperm(x.size(0))
    x_mix = λ*x + (1-λ)*x[idx]
    y_mix = λ*y + (1-λ)*y[idx]
    return x_mix, y_mix

# ───────────────────────── 7. TRAIN LOOP (+ un‑freeze prog) ───────────────
epochs = 80
unf_done=False

for ep in range(epochs):
    # ---------- TRAIN ----------
    model.train(); run=0; p_tr=[]; y_tr=[]
    for i,(x,y) in enumerate(train_loader):
        x,y = x.to(device), y.to(device)
        x,y = mixup(x,y)                                   # MixUp
        opt.zero_grad()
        out = model(x)
        loss = criterion(out,y); loss.backward(); opt.step()
        scheduler.step()
        run += loss.item()*x.size(0)
        p_tr.append(out.detach().cpu()); y_tr.append(y.cpu())

    # ---------- un‑freeze layer3 desde epoch 10 (3) -------
    if ep==9 and not unf_done:
        for n,p in backbone.model.vision_encoder.named_parameters():
            if 'layer3' in n: p.requires_grad=True
        opt.add_param_group({'params':[p for n,p in backbone.model.vision_encoder.named_parameters()
                                       if 'layer3' in n],
                             'lr':lr_back/10})
        unf_done=True

    # ---------- MÉTRICAS TRAIN ----------
    p_tr=torch.cat(p_tr); y_tr=torch.cat(y_tr)
    p_tr_o=torch.tensor(scaler.inverse_transform(p_tr),dtype=torch.float32)
    y_tr_o=torch.tensor(scaler.inverse_transform(y_tr),dtype=torch.float32)
    r2_tr=[r2_score(y_tr_o[:,i],p_tr_o[:,i]) for i in range(11)]
    print(f'Epoch {ep+1}/{epochs} | Train Loss {run/len(train_loader.dataset):.4f}')
    for col,r in zip(label_columns,r2_tr): print(f'  Train R² {col}: {r:.4f}')

    # ---------- VALID ----------
    model.eval(); v_run=0; p_val=[]; y_val=[]
    with torch.no_grad():
        for x,y in val_loader:
            x,y=x.to(device),y.to(device)
            out=model(x)
            v_run+=criterion(out,y).item()*x.size(0)
            p_val.append(out.cpu()); y_val.append(y.cpu())
    p_val=torch.cat(p_val); y_val=torch.cat(y_val)
    p_val_o=torch.tensor(scaler.inverse_transform(p_val),dtype=torch.float32)
    y_val_o=torch.tensor(scaler.inverse_transform(y_val),dtype=torch.float32)
    r2_val=[r2_score(y_val_o[:,i],p_val_o[:,i]) for i in range(11)]
    print(' '*12+f'| Val Loss {v_run/len(val_loader.dataset):.4f}')
    for col,r in zip(label_columns,r2_val): print(f'  Val  R² {col}: {r:.4f}')
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.metrics import r2_score
from reben_publication.BigEarthNetv2_0_ImageClassifier import BigEarthNetv2_0_ImageClassifier

# --- 1. Cargar el backbone preentrenado ---
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

backbone = BigEarthNetv2_0_ImageClassifier.from_pretrained("BIFOLD-BigEarthNetv2-0/resnet50-s2-v0.2.0")
backbone.eval()
backbone = backbone.to(device)

# --- 2. Reemplazar la capa fully-connected (fc) por una identidad ---
backbone.model._modules["vision_encoder"]._modules["fc"] = nn.Identity()

# --- 3. Freeze la mayor parte del backbone y descolar (unfreeze) algunas capas para fine tuning ---
for param in backbone.parameters():
    param.requires_grad = False

# Descongelar (unfreeze) el bloque final (por ejemplo, las capas que contengan "layer4")
for name, param in backbone.model._modules["vision_encoder"].named_parameters():
    if "layer4" in name:
        param.requires_grad = True

# --- Módulo de agregación con atención que utiliza una máscara ---
class AttentionAggregator(nn.Module):
    def __init__(self, feature_dim):
        super(AttentionAggregator, self).__init__()
        # Mapea cada vector de características a un score escalar
        self.attention = nn.Linear(feature_dim, 1)

    def forward(self, x, mask=None):
        """
        x: tensor de forma [batch, seq, feature_dim]
        mask: tensor de forma [batch, seq] con 1 para entradas válidas y 0 para padding.
              Si mask es None, se aplica atención sin enmascarar.
        """
        scores = self.attention(x)              # [batch, seq, 1]
        if mask is not None:
            # Expandir la máscara y asignar -1e9 a posiciones de padding
            mask = mask.unsqueeze(-1)           # [batch, seq, 1]
            scores = scores.masked_fill(mask == 0, -1e9)
        weights = torch.softmax(scores, dim=1)   # [batch, seq, 1]
        aggregated = torch.sum(weights * x, dim=1) # [batch, feature_dim]
        return aggregated

# --- Modelo que integra el backbone, la atención con máscara y capas adicionales ---
class BigEarthNetFineTune(nn.Module):
    def __init__(self, backbone, num_outputs=11, dense_units=1024, dropout_rate=0.3):
        """
        backbone: Modelo BigEarthNet preentrenado (con fc reemplazado por Identity)
        num_outputs: número de salidas (variables continuas a predecir)
        """
        super(BigEarthNetFineTune, self).__init__()
        self.backbone = backbone
        # Módulo de atención para agregar la secuencia
        self.aggregator = AttentionAggregator(feature_dim=2048)
        # Capas fully-connected adicionales para la cabeza del modelo
        self.fc1 = nn.Linear(2048, dense_units)
        self.ln1 = nn.LayerNorm(dense_units)
        self.dropout1 = nn.Dropout(dropout_rate)
        self.fc2 = nn.Linear(dense_units, dense_units // 2)
        self.ln2 = nn.LayerNorm(dense_units // 2)
        self.dropout2 = nn.Dropout(dropout_rate)
        self.out = nn.Linear(dense_units // 2, num_outputs)

    def forward(self, x):
        """
        x: tensor de forma [batch, secuencia, canales, alto, ancho]
        Se asume que las regiones con menos imágenes fueron paddeadas con ceros.
        """
        # Calcular una máscara a partir de la suma absoluta en cada imagen
        # Si la suma es 0, se asume que es padding.
        mask = (x.abs().sum(dim=(2,3,4)) > 0).float()  # [batch, seq]

        batch_size, seq_len, C, H, W = x.shape
        # Fusionar la dimensión de secuencia y batch para procesar cada imagen individualmente
        x = x.view(batch_size * seq_len, C, H, W)
        # Extraer características con el backbone. Salida: [batch*seq, 2048]
        features = self.backbone(x)
        # Regresar a forma secuencial: [batch, seq, 2048]
        features = features.view(batch_size, seq_len, -1)
        # Agregar características a lo largo de la secuencia con atención (usando la máscara)
        aggregated = self.aggregator(features, mask=mask)  # [batch, 2048]
        # Pasar por las capas fully-connected
        x = self.fc1(aggregated)
        x = self.ln1(x)
        x = torch.relu(x)
        x = self.dropout1(x)
        x = self.fc2(x)
        x = self.ln2(x)
        x = torch.relu(x)
        x = self.dropout2(x)
        out = self.out(x)
        return out

# Instanciar el modelo para fine tuning
model_ft = BigEarthNetFineTune(backbone, num_outputs=11, dense_units=1024, dropout_rate=0.3)
model_ft = model_ft.to(device)

# --- Configurar el optimizador con learning rates diferenciales y la función de pérdida ---
optimizer = optim.Adam([
    {'params': model_ft.backbone.parameters(), 'lr': 1e-5},
    {'params': list(model_ft.aggregator.parameters()) +
                list(model_ft.fc1.parameters()) +
                list(model_ft.ln1.parameters()) +
                list(model_ft.dropout1.parameters()) +
                list(model_ft.fc2.parameters()) +
                list(model_ft.ln2.parameters()) +
                list(model_ft.dropout2.parameters()) +
                list(model_ft.out.parameters()),
     'lr': 1e-4}
])
criterion = nn.MSELoss()

# Scheduler para reducir el learning rate según la pérdida de validación
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=3, verbose=True)

# Lista de nombres de etiquetas
label_columns = [
    'educacion_jefe_hogar',
    'hogar_urbano',
    'mujer_jefe_hogar',
    'densidad_hogar',
    'educacion_promedio_hogar',
    'jefe_trabajo_semana_pasada',
    'material_aceptable',
    'material_recuperable',
    'dependencia_hogar',
    'numero_personas',
    'edad_promedio_hogar'
]

train_loss_history = []
val_loss_history = []

# --- Bucle de entrenamiento y validación con cálculo de R² por etiqueta ---
num_epochs = 20


# Cada batch de imágenes debe tener forma: [batch, secuencia, canales, alto, ancho]
# y las etiquetas: [batch, num_outputs]
for epoch in range(num_epochs):
    model_ft.train()
    running_loss = 0.0
    train_preds_list = []
    train_labels_list = []

    for batch_images, batch_labels in train_loader:
        batch_images = batch_images.to(device)  # [batch, seq, C, H, W]
        batch_labels = batch_labels.to(device)  # [batch, num_outputs]

        optimizer.zero_grad()
        outputs = model_ft(batch_images)  # [batch, num_outputs]
        loss = criterion(outputs, batch_labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * batch_images.size(0)
        train_preds_list.append(outputs.detach().cpu())
        train_labels_list.append(batch_labels.detach().cpu())

    epoch_loss = running_loss / len(train_loader.dataset)
    train_loss_history.append(epoch_loss)
    # Concatenar predicciones y etiquetas de entrenamiento
    train_preds = torch.cat(train_preds_list, dim=0)  # [N_train, num_outputs]
    train_labels = torch.cat(train_labels_list, dim=0)  # [N_train, num_outputs]

    # Calcular R² para cada etiqueta en entrenamiento
    train_r2_scores = []
    for i in range(train_preds.shape[1]):
        r2 = r2_score(train_labels[:, i].numpy(), train_preds[:, i].numpy())
        train_r2_scores.append(r2)

    print(f"Epoch {epoch+1}/{num_epochs}, Training Loss: {epoch_loss:.4f}")
    for label, r2 in zip(label_columns, train_r2_scores):
        print(f"  Training R² for {label}: {r2:.4f}")

    # Validación
    model_ft.eval()
    val_loss = 0.0
    val_preds_list = []
    val_labels_list = []

    with torch.no_grad():
        for val_images, val_labels in val_loader:
            val_images = val_images.to(device)
            val_labels = val_labels.to(device)
            outputs = model_ft(val_images)
            loss = criterion(outputs, val_labels)
            val_loss += loss.item() * val_images.size(0)
            val_preds_list.append(outputs.cpu())
            val_labels_list.append(val_labels.cpu())

    val_loss = val_loss / len(val_loader.dataset)
    val_loss_history.append(val_loss)
    # Concatenar predicciones y etiquetas de validación
    val_preds = torch.cat(val_preds_list, dim=0)  # [N_val, num_outputs]
    val_labels = torch.cat(val_labels_list, dim=0)  # [N_val, num_outputs]

    # Calcular R² para cada etiqueta en validación
    val_r2_scores = []
    for i in range(val_preds.shape[1]):
        r2 = r2_score(val_labels[:, i].numpy(), val_preds[:, i].numpy())
        val_r2_scores.append(r2)

    print(f"Epoch {epoch+1}/{num_epochs}, Validation Loss: {val_loss:.4f}")
    for label, r2 in zip(label_columns, val_r2_scores):
        print(f"  Validation R² for {label}: {r2:.4f}")

    # Actualizar scheduler en función de la pérdida de validación
    scheduler.step(val_loss)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.metrics import r2_score
from reben_publication.BigEarthNetv2_0_ImageClassifier import BigEarthNetv2_0_ImageClassifier

# ----- 1. Cargar el backbone preentrenado -----
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
backbone = BigEarthNetv2_0_ImageClassifier.from_pretrained("BIFOLD-BigEarthNetv2-0/resnet50-s2-v0.2.0")
backbone.eval()
backbone = backbone.to(device)
# Reemplazar la capa final para extraer características
backbone.model._modules["vision_encoder"]._modules["fc"] = nn.Identity()
# Congelar la mayoría de los parámetros y descongelar las capas finales
for param in backbone.parameters():
    param.requires_grad = False
for name, param in backbone.model._modules["vision_encoder"].named_parameters():
    if "layer4" in name:
        param.requires_grad = True

# ----- 2. Definir el agregador basado en Set Transformer -----
class SetTransformerAggregator(nn.Module):
    """
    Agregador basado en Set Transformer para combinar un conjunto de vectores de características.
    Se utiliza un bloque de atención para que las interacciones entre imágenes se modelen de forma
    flexible, sin asumir orden en el input.
    """
    def __init__(self, input_dim, num_heads=4, num_inducing=16, embed_dim=256):
        super(SetTransformerAggregator, self).__init__()
        # Parámetro de "puntos inductores" para inicializar las consultas
        self.inducing_points = nn.Parameter(torch.randn(num_inducing, input_dim))
        # Utilizamos MultiheadAttention; configuramos batch_first=True para trabajar con tensores [B, N, dim]
        self.self_attn = nn.MultiheadAttention(embed_dim=input_dim, num_heads=num_heads, batch_first=True)
        self.fc = nn.Linear(input_dim, embed_dim)

    def forward(self, x):
        """
        x: tensor de forma [batch, num_imágenes, input_dim]
        """
        batch_size = x.size(0)
        # Expandir inducing_points a cada muestra del batch
        queries = self.inducing_points.unsqueeze(0).expand(batch_size, -1, -1)  # [B, num_inducing, input_dim]
        # Aplicar atención cruzada: queries se atienden a x (clave y valor)
        attn_output, _ = self.self_attn(queries, x, x)
        # Agregamos los "puntos inductores" con una operación de pooling (por ejemplo, promedio)
        aggregated = attn_output.mean(dim=1)  # [B, input_dim]
        aggregated = self.fc(aggregated)       # [B, embed_dim]
        return aggregated

# ----- 3. Modelo usando Deep Sets / Set Transformer -----
class AdvancedSetModel(nn.Module):
    def __init__(self, backbone, num_outputs=11, dense_units=1024, dropout_rate=0.3,
                 aggregator_input_dim=2048, aggregator_embed_dim=256):
        """
        backbone: Modelo preentrenado (BigEarthNet) que extrae características de cada imagen
        num_outputs: número de salidas (por ejemplo, variables continuas)
        """
        super(AdvancedSetModel, self).__init__()
        self.backbone = backbone
        # Este modelo procesa imágenes con 10 bandas. Se asume que el backbone ha sido adaptado a 10 canales.

        # Agregador para conjuntos, basado en Set Transformer
        self.aggregator = SetTransformerAggregator(input_dim=aggregator_input_dim,
                                                   num_heads=4,
                                                   num_inducing=16,
                                                   embed_dim=aggregator_embed_dim)

        # Cabeza final que procesa la representación agregada
        self.fc1 = nn.Linear(aggregator_embed_dim, dense_units)
        self.ln1 = nn.LayerNorm(dense_units)
        self.dropout1 = nn.Dropout(dropout_rate)
        self.fc2 = nn.Linear(dense_units, dense_units // 2)
        self.ln2 = nn.LayerNorm(dense_units // 2)
        self.dropout2 = nn.Dropout(dropout_rate)
        self.out = nn.Linear(dense_units // 2, num_outputs)

    def forward(self, x):
        """
        x: tensor de forma [batch, conjunto_imágenes, canales, alto, ancho]
        Las imágenes se tratan como un conjunto sin orden.
        """
        batch_size, num_imgs, C, H, W = x.shape
        # Fusionar batch y conjunto para procesar cada imagen individualmente
        x = x.view(batch_size * num_imgs, C, H, W)
        # Extraer características (de dimensión 2048) para cada imagen
        features = self.backbone(x)
        # Recuperar la forma: [batch, num_imgs, 2048]
        features = features.view(batch_size, num_imgs, -1)
        # Agregar el conjunto de características con el Set Transformer
        aggregated = self.aggregator(features)  # [batch, aggregator_embed_dim]
        # Cabeza fully-connected
        x = self.fc1(aggregated)
        x = self.ln1(x)
        x = torch.relu(x)
        x = self.dropout1(x)
        x = self.fc2(x)
        x = self.ln2(x)
        x = torch.relu(x)
        x = self.dropout2(x)
        out = self.out(x)
        return out

# Instanciar el modelo avanzado
model_ft = AdvancedSetModel(backbone, num_outputs=11, dense_units=1024, dropout_rate=0.3,
                            aggregator_input_dim=2048, aggregator_embed_dim=256)
model_ft = model_ft.to(device)

# ----- 4. Configuración del optimizador, la función de pérdida y el scheduler -----
optimizer = optim.Adam([
    {'params': model_ft.backbone.parameters(), 'lr': 1e-5},
    {'params': list(model_ft.aggregator.parameters()) +
               list(model_ft.fc1.parameters()) +
               list(model_ft.ln1.parameters()) +
               list(model_ft.dropout1.parameters()) +
               list(model_ft.fc2.parameters()) +
               list(model_ft.ln2.parameters()) +
               list(model_ft.dropout2.parameters()) +
               list(model_ft.out.parameters()),
     'lr': 1e-4}
])
criterion = nn.MSELoss()
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=3, verbose=True)

# ----- 5. Bucle de entrenamiento y validación -----

# Cada batch de imágenes debe tener forma: [batch, conjunto_imágenes, canales, alto, ancho]
# y las etiquetas: [batch, num_outputs]

num_epochs = 10
label_columns = [
    'educacion_jefe_hogar',
    'hogar_urbano',
    'mujer_jefe_hogar',
    'densidad_hogar',
    'educacion_promedio_hogar',
    'jefe_trabajo_semana_pasada',
    'material_aceptable',
    'material_recuperable',
    'dependencia_hogar',
    'numero_personas',
    'edad_promedio_hogar'
]
train_loss_history = []
val_loss_history = []

for epoch in range(num_epochs):
    model_ft.train()
    running_loss = 0.0
    train_preds_list = []
    train_labels_list = []

    for batch_images, batch_labels in train_loader:
        batch_images = batch_images.to(device)  # [B, num_imgs, C, H, W]
        batch_labels = batch_labels.to(device)  # [B, num_outputs]

        optimizer.zero_grad()
        outputs = model_ft(batch_images)        # [B, num_outputs]
        loss = criterion(outputs, batch_labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * batch_images.size(0)
        train_preds_list.append(outputs.detach().cpu())
        train_labels_list.append(batch_labels.detach().cpu())

    epoch_loss = running_loss / len(train_loader.dataset)
    train_loss_history.append(epoch_loss)

    # Concatenar predicciones y etiquetas de entrenamiento para calcular R²
    train_preds = torch.cat(train_preds_list, dim=0)
    train_labels = torch.cat(train_labels_list, dim=0)
    train_r2_scores = []
    for i in range(train_preds.shape[1]):
        r2 = r2_score(train_labels[:, i].numpy(), train_preds[:, i].numpy())
        train_r2_scores.append(r2)

    print(f"Epoch {epoch+1}/{num_epochs}, Training Loss: {epoch_loss:.4f}")
    for label, r2 in zip(label_columns, train_r2_scores):
        print(f"  Training R² for {label}: {r2:.4f}")

    # Validación
    model_ft.eval()
    val_loss = 0.0
    val_preds_list = []
    val_labels_list = []
    with torch.no_grad():
        for val_images, val_labels in val_loader:
            val_images = val_images.to(device)
            val_labels = val_labels.to(device)
            outputs = model_ft(val_images)
            loss = criterion(outputs, val_labels)
            val_loss += loss.item() * val_images.size(0)
            val_preds_list.append(outputs.cpu())
            val_labels_list.append(val_labels.cpu())

    val_loss = val_loss / len(val_loader.dataset)
    val_loss_history.append(val_loss)

    val_preds = torch.cat(val_preds_list, dim=0)
    val_labels = torch.cat(val_labels_list, dim=0)
    val_r2_scores = []
    for i in range(val_preds.shape[1]):
        r2 = r2_score(val_labels[:, i].numpy(), val_preds[:, i].numpy())
        val_r2_scores.append(r2)

    print(f"Epoch {epoch+1}/{num_epochs}, Validation Loss: {val_loss:.4f}")
    for label, r2 in zip(label_columns, val_r2_scores):
        print(f"  Validation R² for {label}: {r2:.4f}")

    scheduler.step(val_loss)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from sklearn.metrics import r2_score
from reben_publication.BigEarthNetv2_0_ImageClassifier import BigEarthNetv2_0_ImageClassifier

# =============================================================================
# 1. Cargar el Backbone Preentrenado y Configuración Base
# =============================================================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Se carga el modelo BigEarthNet preentrenado. Se asume que trabaja con 10 bandas.
backbone = BigEarthNetv2_0_ImageClassifier.from_pretrained("BIFOLD-BigEarthNetv2-0/resnet50-s2-v0.2.0")
backbone.eval()
backbone = backbone.to(device)

# Reemplazar la capa final (fully-connected) por una identidad para extraer características
backbone.model._modules["vision_encoder"]._modules["fc"] = nn.Identity()

# Congelar los parámetros de la mayoría del backbone y dejar afinables las últimas capas (por ejemplo, de layer4)
for param in backbone.parameters():
    param.requires_grad = False
for name, param in backbone.model._modules["vision_encoder"].named_parameters():
    if "layer4" in name:
        param.requires_grad = True

# =============================================================================
# 2. Módulo de Fusión Multi-Escala con Atención
# =============================================================================
class MultiScaleAttentionFusion(nn.Module):
    """
    Fusiona características extraídas de diferentes escalas (por ejemplo, de layer1, layer2, layer3 y layer4)
    usando convoluciones 1x1 para estandarizar el número de canales y una subred de atención que pondera cada escala.
    """
    def __init__(self, in_channels_list, out_channels):
        """
        :param in_channels_list: lista con la cantidad de canales de cada escala.
                                 Ejemplo típico para ResNet50: [256, 512, 1024, 2048]
        :param out_channels: número de canales con el que se transformarán todas las escalas.
        """
        super(MultiScaleAttentionFusion, self).__init__()
        # Convertir cada escala a la misma cantidad de canales
        self.conv_layers = nn.ModuleList([
            nn.Conv2d(in_ch, out_channels, kernel_size=1)
            for in_ch in in_channels_list
        ])
        # Módulo de atención: se concatena la salida de las transformaciones y se genera un mapa de pesos
        self.attention = nn.Sequential(
            nn.Conv2d(out_channels * len(in_channels_list), out_channels, kernel_size=1),
            nn.ReLU(),
            nn.Conv2d(out_channels, len(in_channels_list), kernel_size=1),
            nn.Softmax(dim=1)
        )

    def forward(self, features):
        """
        :param features: lista de tensores [B, C_i, H_i, W_i] extraídos de distintas escalas.
        :return: tensor fusionado [B, out_channels, H, W] (donde H, W corresponden a la resolución de la escala de referencia)
        """
        # Transformar cada entrada con su respectiva convolución
        features_transformed = [conv(feat) for conv, feat in zip(self.conv_layers, features)]
        # Tomamos como referencia la resolución de la primera escala (generalmente la de mayor resolución)
        target_size = features_transformed[0].shape[2:]
        # Reescalar (upsample) todas las características para que tengan la misma resolución
        features_aligned = [
            F.interpolate(feat, size=target_size, mode='bilinear', align_corners=False)
            for feat in features_transformed
        ]
        # Concatenar a lo largo de la dimensión de canales
        concatenated = torch.cat(features_aligned, dim=1)  # [B, out_channels * num_scales, H, W]
        # Calcular pesos de atención sobre cada escala
        attention_weights = self.attention(concatenated)  # [B, num_scales, H, W]
        # Aplicar los pesos a cada escala de forma independiente
        weighted_features = [
            feat * attention_weights[:, i:i+1, :, :]
            for i, feat in enumerate(features_aligned)
        ]
        # Fusionar las escalas mediante suma
        fused = torch.sum(torch.stack(weighted_features, dim=0), dim=0)  # [B, out_channels, H, W]
        return fused

# =============================================================================
# 3. Extracción Multi-Escala en el Backbone
# =============================================================================
class MultiScaleBackbone(nn.Module):
    """
    Envuelve el backbone preentrenado para extraer características de distintas escalas.
    Se asume que el backbone es de tipo ResNet y se extraen las salidas de:
      - layer1 (por ejemplo, 256 canales)
      - layer2 (por ejemplo, 512 canales)
      - layer3 (por ejemplo, 1024 canales)
      - layer4 (por ejemplo, 2048 canales)
    Luego se fusionan mediante un módulo de fusión.
    """
    def __init__(self, backbone, fuse_module):
        super(MultiScaleBackbone, self).__init__()
        self.backbone = backbone
        self.fuse_module = fuse_module

    def forward(self, x):
        # x: [B, C, H, W]
        encoder = self.backbone.model._modules["vision_encoder"]
        # Procesar etapa inicial (conv1, bn1, activación y maxpool)
        x = encoder.conv1(x)   # [B, 64, H/2, W/2]
        x = encoder.bn1(x)
        x = F.relu(x)          # <-- Usamos F.relu en lugar de encoder.relu(x)
        x = encoder.maxpool(x) # [B, 64, H/4, W/4]

        # Extraer salidas de distintos bloques
        feat1 = encoder.layer1(x)  # Ej: [B, 256, H/4, W/4]
        feat2 = encoder.layer2(feat1)  # Ej: [B, 512, H/8, W/8]
        feat3 = encoder.layer3(feat2)  # Ej: [B, 1024, H/16, W/16]
        feat4 = encoder.layer4(feat3)  # Ej: [B, 2048, H/32, W/32]

        # Fusionar las características extraídas de distintos niveles
        fused = self.fuse_module([feat1, feat2, feat3, feat4])
        # Aplicar Global Average Pooling para obtener un vector por imagen
        gap = F.adaptive_avg_pool2d(fused, (1, 1))
        gap = gap.view(gap.size(0), -1)  # [B, out_channels]
        return gap

# =============================================================================
# 4. Agregador Basado en Set Transformer
# =============================================================================
class SetTransformerAggregator(nn.Module):
    """
    Agrega un conjunto de vectores (uno por imagen de la región) usando mecanismos de atención.
    Permite ajustar el número de puntos inductores y las cabezas de atención para capturar
    relaciones complejas entre las imágenes del set.
    """
    def __init__(self, input_dim, num_heads=4, num_inducing=16, embed_dim=256):
        super(SetTransformerAggregator, self).__init__()
        self.inducing_points = nn.Parameter(torch.randn(num_inducing, input_dim))
        self.self_attn = nn.MultiheadAttention(embed_dim=input_dim, num_heads=num_heads, batch_first=True)
        self.fc = nn.Linear(input_dim, embed_dim)

    def forward(self, x):
        # x: [B, num_imágenes, input_dim]
        batch_size = x.size(0)
        # Expandir los inducing points para cada muestra en el batch
        queries = self.inducing_points.unsqueeze(0).expand(batch_size, -1, -1)  # [B, num_inducing, input_dim]
        attn_output, _ = self.self_attn(queries, x, x)
        # Promediar sobre los inducing points
        aggregated = attn_output.mean(dim=1)  # [B, input_dim]
        aggregated = self.fc(aggregated)       # [B, embed_dim]
        return aggregated

# =============================================================================
# 5. Modelo: Advanced Multi-Scale Set Transformer
# =============================================================================
class AdvancedMultiScaleSetModel(nn.Module):
    def __init__(self, backbone, num_outputs=11, dense_units=2048, dropout_rate=0.2,
                 num_heads=8, num_inducing=16):
        """
        :param backbone: modelo preentrenado (BigEarthNet) que se modificará para extraer características multi-escala
        :param num_outputs: número de variables a predecir
        :param dense_units: unidades de la capa densa intermedia
        :param dropout_rate: tasa de dropout para regularización
        :param num_heads: número de cabezas en el agregador del set transformer
        :param num_inducing: número de inducing points en el agregador del set transformer
        """
        super(AdvancedMultiScaleSetModel, self).__init__()
        # Definir módulo de fusión multi-escala
        self.fuse_module = MultiScaleAttentionFusion(in_channels_list=[256, 512, 1024, 2048],
                                                     out_channels=256)
        # Envolver el backbone para obtener un extractor de características multi-escala
        self.multi_scale_backbone = MultiScaleBackbone(backbone, self.fuse_module)
        # Agregador del set transformer (para combinar la información de cada imagen en el conjunto)
        self.aggregator = SetTransformerAggregator(input_dim=256, num_heads=num_heads,
                                                   num_inducing=num_inducing, embed_dim=256)
        # Cabeza final (fully connected) para la predicción
        self.fc1 = nn.Linear(256, dense_units)
        self.ln1 = nn.LayerNorm(dense_units)
        self.dropout1 = nn.Dropout(dropout_rate)
        self.fc2 = nn.Linear(dense_units, dense_units // 2)
        self.ln2 = nn.LayerNorm(dense_units // 2)
        self.dropout2 = nn.Dropout(dropout_rate)
        self.out = nn.Linear(dense_units // 2, num_outputs)

    def forward(self, x):
        """
        :param x: tensor de forma [B, num_imágenes, canales, alto, ancho]
                  Cada conjunto representa una región sin orden intrínseco.
        """
        B, num_imgs, C, H, W = x.shape
        # Procesar cada imagen individualmente: fusiona las escalas y obtiene un vector de dimensión 256
        x = x.view(B * num_imgs, C, H, W)
        features = self.multi_scale_backbone(x)  # [B*num_imgs, 256]
        # Reorganizar para obtener el set: [B, num_imgs, 256]
        features = features.view(B, num_imgs, -1)
        # Agregar la información del set usando el set transformer aggregator
        aggregated = self.aggregator(features)  # [B, 256]
        # Cabeza fully-connected para la predicción final
        x = self.fc1(aggregated)
        x = self.ln1(x)
        x = F.relu(x)
        x = self.dropout1(x)
        x = self.fc2(x)
        x = self.ln2(x)
        x = F.relu(x)
        x = self.dropout2(x)
        out = self.out(x)
        return out

# =============================================================================
# 6. Instanciar el Modelo y Configurar el Entrenamiento
# =============================================================================
model_ft = AdvancedMultiScaleSetModel(backbone, num_outputs=11, dense_units=2048, dropout_rate=0.2,
                                      num_heads=8, num_inducing=16)
model_ft = model_ft.to(device)

optimizer = optim.Adam([
    {'params': model_ft.multi_scale_backbone.parameters(), 'lr': 1e-4},
    {'params': list(model_ft.aggregator.parameters()) +
               list(model_ft.fc1.parameters()) +
               list(model_ft.ln1.parameters()) +
               list(model_ft.dropout1.parameters()) +
               list(model_ft.fc2.parameters()) +
               list(model_ft.ln2.parameters()) +
               list(model_ft.dropout2.parameters()) +
               list(model_ft.out.parameters()),
     'lr': 1e-4}
])
criterion = nn.MSELoss()
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.8, patience=6, verbose=True)

# =============================================================================
# 7. Bucle de Entrenamiento y Validación
# =============================================================================
num_epochs = 60
label_columns = [
    'educacion_jefe_hogar',
    'hogar_urbano',
    'mujer_jefe_hogar',
    'densidad_hogar',
    'educacion_promedio_hogar',
    'jefe_trabajo_semana_pasada',
    'material_aceptable',
    'material_recuperable',
    'dependencia_hogar',
    'numero_personas',
    'edad_promedio_hogar'
]
train_loss_history = []
val_loss_history = []

for epoch in range(num_epochs):
    model_ft.train()
    running_loss = 0.0
    train_preds_list = []
    train_labels_list = []

    for batch_images, batch_labels in train_loader:
        # batch_images: [B, num_imágenes, canales, alto, ancho]
        # batch_labels: [B, num_outputs]
        batch_images = batch_images.to(device)
        batch_labels = batch_labels.to(device)

        optimizer.zero_grad()
        outputs = model_ft(batch_images)
        loss = criterion(outputs, batch_labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * batch_images.size(0)
        train_preds_list.append(outputs.detach().cpu())
        train_labels_list.append(batch_labels.detach().cpu())

    epoch_loss = running_loss / len(train_loader.dataset)
    train_loss_history.append(epoch_loss)

    # Calcular R² para entrenamiento
    train_preds = torch.cat(train_preds_list, dim=0)
    train_labels = torch.cat(train_labels_list, dim=0)
    train_r2_scores = []
    for i in range(train_preds.shape[1]):
        r2 = r2_score(train_labels[:, i].numpy(), train_preds[:, i].numpy())
        train_r2_scores.append(r2)

    print(f"Epoch {epoch+1}/{num_epochs}, Training Loss: {epoch_loss:.4f}")
    for label, r2 in zip(label_columns, train_r2_scores):
        print(f"  Training R² for {label}: {r2:.4f}")

    # Validación
    model_ft.eval()
    val_loss = 0.0
    val_preds_list = []
    val_labels_list = []
    with torch.no_grad():
        for val_images, val_labels in val_loader:
            val_images = val_images.to(device)
            val_labels = val_labels.to(device)
            outputs = model_ft(val_images)
            loss = criterion(outputs, val_labels)
            val_loss += loss.item() * val_images.size(0)
            val_preds_list.append(outputs.cpu())
            val_labels_list.append(val_labels.cpu())

    val_loss = val_loss / len(val_loader.dataset)
    val_loss_history.append(val_loss)

    val_preds = torch.cat(val_preds_list, dim=0)
    val_labels = torch.cat(val_labels_list, dim=0)
    val_r2_scores = []
    for i in range(val_preds.shape[1]):
        r2 = r2_score(val_labels[:, i].numpy(), val_preds[:, i].numpy())
        val_r2_scores.append(r2)

    print(f"Epoch {epoch+1}/{num_epochs}, Validation Loss: {val_loss:.4f}")
    for label, r2 in zip(label_columns, val_r2_scores):
        print(f"  Validation R² for {label}: {r2:.4f}")

    scheduler.step(val_loss)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from sklearn.metrics import r2_score
from reben_publication.BigEarthNetv2_0_ImageClassifier import BigEarthNetv2_0_ImageClassifier

# =============================================================================
# 0. Definir la función de pérdida multi-tarea personalizada
# =============================================================================
class MultiTaskLoss(nn.Module):
    def __init__(self, gamma=2.0):
        """
        :param gamma: parámetro para la focal loss en regresión (ajústalo según necesites)
        """
        super(MultiTaskLoss, self).__init__()
        self.gamma = gamma
        self.mse_loss = nn.MSELoss(reduction='mean')
        self.bce_loss = nn.BCELoss(reduction='mean')
        self.huber_loss = nn.SmoothL1Loss(reduction='mean')  # Aproximación al Huber Loss

    def focal_loss(self, input, target):
        """
        Focal loss simple para regresión:
          focal_loss = |error|^gamma * (error)^2
        """
        error = input - target
        loss = (torch.abs(error) ** self.gamma) * (error ** 2)
        return loss.mean()

    def forward(self, outputs, targets):
        """
        Se espera que outputs y targets tengan forma [batch_size, 11] y que:
          Índice 0 (educacion_jefe_hogar): MSE.
          Índice 1 (hogar_urbano): BCE (se aplica sigmoid).
          Índice 2 (mujer_jefe_hogar): MSE.
          Índice 3 (densidad_hogar): log1p + Huber Loss.
          Índice 4 (educacion_promedio_hogar): MSE.
          Índice 5 (jefe_trabajo_semana_pasada): MSE.
          Índice 6 (material_aceptable): MSE.
          Índice 7 (material_recuperable): Focal Loss.
          Índice 8 (dependencia_hogar): log1p + Huber Loss.
          Índice 9 (numero_personas): MSE.
          Índice 10 (edad_promedio_hogar): MSE.
        """
        total_loss = 0.0
        # Variables "normales" con MSE: índices 0, 4, 5, 6, 9, 10
        for idx in [0, 4, 5, 6, 9, 10]:
            total_loss += self.mse_loss(outputs[:, idx], targets[:, idx])

        # Variable binaria (hogar_urbano): índice 1
        # Se aplica sigmoid para obtener probabilidades en [0,1]
        total_loss += self.mse_loss(outputs[:, 1], targets[:, 1])

        # Variable sesgada (material_recuperable): índice 7 (Focal Loss)
        total_loss += self.focal_loss(outputs[:, 7], targets[:, 7])

        # Variables concentradas (densidad_hogar y dependencia_hogar): índices 3 y 8
        for idx in [3, 8]:
            # Se aplica log1p para "aplanar" la distribución (clamp para evitar valores negativos)
            out_log = torch.log1p(torch.clamp(outputs[:, idx], min=0))
            tgt_log = torch.log1p(torch.clamp(targets[:, idx], min=0))
            total_loss += self.huber_loss(out_log, tgt_log)

        # Variable restante: mujer_jefe_hogar (índice 2)
        total_loss += self.mse_loss(outputs[:, 2], targets[:, 2])

        return total_loss

# =============================================================================
# 1. Cargar el Backbone Preentrenado y Configuración Base
# =============================================================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Se carga el modelo BigEarthNet preentrenado. Se asume que trabaja con 10 bandas.
backbone = BigEarthNetv2_0_ImageClassifier.from_pretrained("BIFOLD-BigEarthNetv2-0/resnet50-s2-v0.2.0")
backbone.eval()
backbone = backbone.to(device)

# Reemplazar la capa final (fully-connected) por una identidad para extraer características
backbone.model._modules["vision_encoder"]._modules["fc"] = nn.Identity()

# Congelar los parámetros de la mayoría del backbone y dejar afinables las últimas capas (por ejemplo, layer4)
for param in backbone.parameters():
    param.requires_grad = False
for name, param in backbone.model._modules["vision_encoder"].named_parameters():
    if "layer4" in name:
        param.requires_grad = True

# =============================================================================
# 2. Módulo de Fusión Multi-Escala con Atención
# =============================================================================
class MultiScaleAttentionFusion(nn.Module):
    def __init__(self, in_channels_list, out_channels):
        super(MultiScaleAttentionFusion, self).__init__()
        self.conv_layers = nn.ModuleList([
            nn.Conv2d(in_ch, out_channels, kernel_size=1)
            for in_ch in in_channels_list
        ])
        self.attention = nn.Sequential(
            nn.Conv2d(out_channels * len(in_channels_list), out_channels, kernel_size=1),
            nn.ReLU(),
            nn.Conv2d(out_channels, len(in_channels_list), kernel_size=1),
            nn.Softmax(dim=1)
        )

    def forward(self, features):
        features_transformed = [conv(feat) for conv, feat in zip(self.conv_layers, features)]
        target_size = features_transformed[0].shape[2:]
        features_aligned = [
            F.interpolate(feat, size=target_size, mode='bilinear', align_corners=False)
            for feat in features_transformed
        ]
        concatenated = torch.cat(features_aligned, dim=1)
        attention_weights = self.attention(concatenated)
        weighted_features = [
            feat * attention_weights[:, i:i+1, :, :]
            for i, feat in enumerate(features_aligned)
        ]
        fused = torch.sum(torch.stack(weighted_features, dim=0), dim=0)
        return fused

# =============================================================================
# 3. Extracción Multi-Escala en el Backbone
# =============================================================================
class MultiScaleBackbone(nn.Module):
    def __init__(self, backbone, fuse_module):
        super(MultiScaleBackbone, self).__init__()
        self.backbone = backbone
        self.fuse_module = fuse_module

    def forward(self, x):
        encoder = self.backbone.model._modules["vision_encoder"]
        x = encoder.conv1(x)
        x = encoder.bn1(x)
        x = F.relu(x)
        x = encoder.maxpool(x)
        feat1 = encoder.layer1(x)
        feat2 = encoder.layer2(feat1)
        feat3 = encoder.layer3(feat2)
        feat4 = encoder.layer4(feat3)
        fused = self.fuse_module([feat1, feat2, feat3, feat4])
        gap = F.adaptive_avg_pool2d(fused, (1, 1))
        gap = gap.view(gap.size(0), -1)
        return gap

# =============================================================================
# 4. Agregador Basado en Set Transformer
# =============================================================================
class SetTransformerAggregator(nn.Module):
    def __init__(self, input_dim, num_heads=8, num_inducing=16, embed_dim=256):
        super(SetTransformerAggregator, self).__init__()
        self.inducing_points = nn.Parameter(torch.randn(num_inducing, input_dim))
        self.self_attn = nn.MultiheadAttention(embed_dim=input_dim, num_heads=num_heads, batch_first=True)
        self.fc = nn.Linear(input_dim, embed_dim)

    def forward(self, x):
        batch_size = x.size(0)
        queries = self.inducing_points.unsqueeze(0).expand(batch_size, -1, -1)
        attn_output, _ = self.self_attn(queries, x, x)
        aggregated = attn_output.mean(dim=1)
        aggregated = self.fc(aggregated)
        return aggregated

# =============================================================================
# 5. Modelo: Advanced Multi-Scale Set Transformer
# =============================================================================
class AdvancedMultiScaleSetModel(nn.Module):
    def __init__(self, backbone, num_outputs=11, dense_units=2048, dropout_rate=0.2,
                 num_heads=8, num_inducing=16):
        super(AdvancedMultiScaleSetModel, self).__init__()
        self.fuse_module = MultiScaleAttentionFusion(in_channels_list=[256, 512, 1024, 2048],
                                                     out_channels=256)
        self.multi_scale_backbone = MultiScaleBackbone(backbone, self.fuse_module)
        self.aggregator = SetTransformerAggregator(input_dim=256, num_heads=num_heads,
                                                   num_inducing=num_inducing, embed_dim=256)
        self.fc1 = nn.Linear(256, dense_units)
        self.ln1 = nn.LayerNorm(dense_units)
        self.dropout1 = nn.Dropout(dropout_rate)
        self.fc2 = nn.Linear(dense_units, dense_units // 2)
        self.ln2 = nn.LayerNorm(dense_units // 2)
        self.dropout2 = nn.Dropout(dropout_rate)
        self.out = nn.Linear(dense_units // 2, num_outputs)

    def forward(self, x):
        B, num_imgs, C, H, W = x.shape
        x = x.view(B * num_imgs, C, H, W)
        features = self.multi_scale_backbone(x)
        features = features.view(B, num_imgs, -1)
        aggregated = self.aggregator(features)
        x = self.fc1(aggregated)
        x = self.ln1(x)
        x = F.relu(x)
        x = self.dropout1(x)
        x = self.fc2(x)
        x = self.ln2(x)
        x = F.relu(x)
        x = self.dropout2(x)
        out = self.out(x)
        return out

# =============================================================================
# 6. Instanciar el Modelo y Configurar el Entrenamiento
# =============================================================================
model_ft = AdvancedMultiScaleSetModel(backbone, num_outputs=11, dense_units=2048, dropout_rate=0.3,
                                      num_heads=8, num_inducing=16)
model_ft = model_ft.to(device)

optimizer = optim.Adam([
    {'params': model_ft.multi_scale_backbone.parameters(), 'lr': 1e-4},
    {'params': list(model_ft.aggregator.parameters()) +
               list(model_ft.fc1.parameters()) +
               list(model_ft.ln1.parameters()) +
               list(model_ft.dropout1.parameters()) +
               list(model_ft.fc2.parameters()) +
               list(model_ft.ln2.parameters()) +
               list(model_ft.dropout2.parameters()) +
               list(model_ft.out.parameters()),
     'lr': 1e-4}
])
# Función de pérdida multi-tarea.
criterion = MultiTaskLoss(gamma=2.0)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.8, patience=6, verbose=True)

# =============================================================================
# 7. Bucle de Entrenamiento y Validación
# =============================================================================
num_epochs = 60
label_columns = [
    'educacion_jefe_hogar',
    'hogar_urbano',
    'mujer_jefe_hogar',
    'densidad_hogar',
    'educacion_promedio_hogar',
    'jefe_trabajo_semana_pasada',
    'material_aceptable',
    'material_recuperable',
    'dependencia_hogar',
    'numero_personas',
    'edad_promedio_hogar'
]
train_loss_history = []
val_loss_history = []

for epoch in range(num_epochs):
    model_ft.train()
    running_loss = 0.0
    train_preds_list = []
    train_labels_list = []

    for batch_images, batch_labels in train_loader:
        batch_images = batch_images.to(device)
        batch_labels = batch_labels.to(device)

        optimizer.zero_grad()
        outputs = model_ft(batch_images)
        loss = criterion(outputs, batch_labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * batch_images.size(0)
        train_preds_list.append(outputs.detach().cpu())
        train_labels_list.append(batch_labels.detach().cpu())

    epoch_loss = running_loss / len(train_loader.dataset)
    train_loss_history.append(epoch_loss)

    train_preds = torch.cat(train_preds_list, dim=0)
    train_labels = torch.cat(train_labels_list, dim=0)
    train_r2_scores = []
    for i in range(train_preds.shape[1]):
        r2 = r2_score(train_labels[:, i].numpy(), train_preds[:, i].numpy())
        train_r2_scores.append(r2)

    print(f"Epoch {epoch+1}/{num_epochs}, Training Loss: {epoch_loss:.4f}")
    for label, r2 in zip(label_columns, train_r2_scores):
        print(f"  Training R² for {label}: {r2:.4f}")

    model_ft.eval()
    val_loss = 0.0
    val_preds_list = []
    val_labels_list = []
    with torch.no_grad():
        for val_images, val_labels in val_loader:
            val_images = val_images.to(device)
            val_labels = val_labels.to(device)
            outputs = model_ft(val_images)
            loss = criterion(outputs, val_labels)
            val_loss += loss.item() * val_images.size(0)
            val_preds_list.append(outputs.cpu())
            val_labels_list.append(val_labels.cpu())

    val_loss = val_loss / len(val_loader.dataset)
    val_loss_history.append(val_loss)

    val_preds = torch.cat(val_preds_list, dim=0)
    val_labels = torch.cat(val_labels_list, dim=0)
    val_r2_scores = []
    for i in range(val_preds.shape[1]):
        r2 = r2_score(val_labels[:, i].numpy(), val_preds[:, i].numpy())
        val_r2_scores.append(r2)

    print(f"Epoch {epoch+1}/{num_epochs}, Validation Loss: {val_loss:.4f}")
    for label, r2 in zip(label_columns, val_r2_scores):
        print(f"  Validation R² for {label}: {r2:.4f}")

    scheduler.step(val_loss)
```
